# 1. Project Context and Objectives

## 1.1 Project Context

We will focus on implementing the Semantic Retrieval layer of the Local Mini RAG Document QA System.

By the end of embeddings and vector database, the document processing and vector indexing infrastructure had been established:

- 228 document chunks were generated using the Recursive Markdown chunking strategy.
- Each chunk was represented by a 768-dimensional embedding.
- The embedding model is `intfloat/multilingual-e5-base`.
- The model uses `passage: ` as the document embedding prefix.
- Embeddings are normalized and stored as `float32`.
- The embeddings were indexed in a persistent ChromaDB collection.
- The collection name is `hadota_chunks_e5_base_v1`.
- The collection uses cosine distance.

The existing infrastructure provides the foundation for semantic retrieval. However, a query-processing and retrieval layer is still required to allow users to search the indexed knowledge base.

Day 4 will extend the existing architecture without regenerating embeddings or modifying the indexed data.

## 1.2 Problem Definition

The system needs a retrieval component that accepts a natural-language query, converts it into a vector representation using the existing embedding model, and searches the existing ChromaDB collection for semantically relevant document chunks.

Unlike traditional keyword matching, semantic retrieval compares vector representations to identify content with similar meaning, even when the query and document use different wording.

The retrieval component must return ranked results containing the retrieved chunk identifiers, document identifiers, text content, metadata, and similarity distances.

## 1.3 Main Objectives

The primary objectives of semantic retrieval:

1. Validate and normalize user queries without changing their semantic meaning.
2. Prepare queries using the required `query: ` prefix.
3. Generate query embeddings using the existing embedding model and configuration.
4. Validate query embedding dimensions, numerical values, and normalization.
5. Search the existing ChromaDB collection using query embeddings.
6. Retrieve the Top-K most relevant document chunks.
7. Return structured and validated retrieval results.
8. Handle invalid inputs, empty results, and retrieval errors.
9. Test retrieval behavior using English, Arabic, and mixed-language queries.
10. Document experimental observations and retrieval limitations.

## 1.4 Scope and Constraints

### In Scope

- Query validation and preparation.
- Query embedding generation.
- Semantic similarity search.
- ChromaDB collection access and retrieval.
- Top-K result selection.
- Structured retrieval result models.
- Unit and integration testing.
- Multilingual retrieval experiments.
- Retrieval reporting and analysis.

### Out of Scope

- Document ingestion and text extraction.
- Document chunking or chunking strategy changes.
- Embedding regeneration.
- Vector database re-indexing or resetting.
- BM25 and hybrid retrieval.
- Reranking models.
- Query rewriting, multi-query retrieval, and HyDE.
- LLM-based answer generation.
- Production deployment and performance claims.

## 1.5 Architectural Principle

Semantic retrieval is an extension of the existing embeddings infrastructure, not a replacement for it.

The implementation must reuse:

- The existing embedding model and pinned revision.
- The existing embedding configuration.
- The existing query embedding interface.
- The existing ChromaDB persistent storage.
- The existing collection and its cosine distance configuration.

No new embedding model or vector database should be introduced.

The retrieval implementation must not modify, regenerate, or re-index the existing document embeddings.

## 1.6 Expected Outcome

At the end, the system should provide a reusable semantic retrieval interface that accepts a natural-language query and returns up to K ranked document chunks from the existing knowledge base.

Each result should contain:

- `chunk_id`
- `document_id`
- `text`
- `metadata`
- `distance`

The returned distance represents the vector distance reported by the configured retrieval backend. It must not be interpreted as a probability or a calibrated confidence score.
