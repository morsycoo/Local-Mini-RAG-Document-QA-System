# Environment & Dependency Audit

## 0.1 Objective

The objective of this section is to inspect the existing Python environment and verify the dependencies and hardware capabilities required for local embedding generation and persistent vector storage.

No package installation, upgrade, downgrade, or environment modification will be performed before the initial audit is completed.

## 0.2 Audit Scope

The initial audit covers:

- Active Python interpreter and version.
- Project virtual environment.
- PyTorch installation and version.
- CUDA availability through PyTorch.
- CUDA runtime version reported by PyTorch.
- NVIDIA GPU detection and available GPU memory.
- Sentence Transformers installation.
- Transformers installation.
- ChromaDB installation.
- Python dependency consistency.

## 0.3 Required Dependencies

| Component | Purpose |
|---|---|
| Python | Project execution environment |
| PyTorch | Neural network inference |
| CUDA | GPU acceleration when available |
| Sentence Transformers | Loading and encoding with E5 |
| Transformers | Tokenizer and model infrastructure |
| ChromaDB | Persistent local vector storage |

## 0.4 Environment Safety Rules

1. Inspect the active interpreter before installing packages.
2. Preserve an existing working PyTorch installation.
3. Do not replace the CUDA-enabled PyTorch build without investigating the dependency resolver output.
4. Install only dependencies that are confirmed to be missing.
5. Use CPU execution as a supported fallback when CUDA is unavailable.
6. Record the actual package versions and hardware environment used in the experiment.

## 0.5 Expected Outcome

A verified environment report containing the active Python interpreter, installed dependency versions, PyTorch configuration, CUDA availability, and detected GPU capabilities.

The results will determine whether the current environment is ready for loading `intfloat/multilingual-e5-base` or requires a controlled dependency adjustment.
