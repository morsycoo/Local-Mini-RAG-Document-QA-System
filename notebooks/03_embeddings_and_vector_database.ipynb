## 1.1 Load `chunks.jsonl`

The primary input for the embedding pipeline is the existing `chunks.jsonl` artifact generated.

The corpus was previously processed using the `recursive_markdown` chunking strategy.

In this step, the existing JSONL file is loaded into memory without modifying or rebuilding any chunks.

The loading procedure includes:

1. Locate the project root directory.
2. Resolve the existing `chunks.jsonl` path.
3. Read each non-empty line as an independent JSON record.
4. Store the loaded records in `raw_chunk_records`.
5. Report the number of loaded records and inspect the first record's field names.

No schema validation, relationship verification, deduplication, or content modification is performed at this stage.


````python
import json
from pathlib import Path


# --------------------------------------------------
# 1. Locate Project Root
# --------------------------------------------------

current_dir = Path.cwd().resolve()

candidate_roots = [current_dir, *current_dir.parents]

matching_roots = [
    root
    for root in candidate_roots
    if (root / "data" / "knowledge_base" / "processed" / "chunks.jsonl").is_file()
]

if len(matching_roots) != 1:
    raise RuntimeError(
        "Could not uniquely identify the project root.\n"
        f"Current working directory: {current_dir}\n"
        f"Matching project roots: {matching_roots}"
    )

PROJECT_ROOT = matching_roots[0]

# --------------------------------------------------
# 2. Resolve Chunks Artifact Path
# --------------------------------------------------

chunks_path = PROJECT_ROOT / "data" / "knowledge_base" / "processed" / "chunks.jsonl"

print("Project root:", PROJECT_ROOT)
print("Chunks file:", chunks_path)
print("File exists:", chunks_path.is_file())

# --------------------------------------------------
# 3. Load JSONL Records
# --------------------------------------------------

raw_chunk_records = []
blank_lines = 0

with chunks_path.open("r", encoding="utf-8-sig") as file:
    for line_number, line in enumerate(file, start=1):
        if not line.strip():
            blank_lines += 1
            continue

        try:
            record = json.loads(line)

        except json.JSONDecodeError as exc:
            raise ValueError(
                f"Invalid JSON at line {line_number} in {chunks_path.name}: {exc}"
            ) from exc

        raw_chunk_records.append(record)

# --------------------------------------------------
# 4. Loading Summary
# --------------------------------------------------

print("\n" + "=" * 55)
print("CHUNKS LOADING SUMMARY")
print("=" * 55)

print("Loaded records:", len(raw_chunk_records))
print("Blank lines skipped:", blank_lines)
print(
    "First record type:",
    type(raw_chunk_records[0]).__name__ if raw_chunk_records else "No records",
)

if raw_chunk_records and isinstance(raw_chunk_records[0], dict):
    print("First record fields:", list(raw_chunk_records[0].keys()))

print("=" * 55)
````

### Result Interpretation

The `chunks.jsonl` artifact was located and loaded successfully.

The loading operation returned:

- Total records loaded: 228
- Blank lines skipped: 0
- First record type: `dict`
- JSON parsing errors: 0

The first record contains 13 fields:

`chunk_id`, `document_id`, `page_number`, `chunk_index`, `content`, `char_start`, `char_end`, `content_hash`, `section_path`, `strategy`, `config_signature`, `chunking_version`, and `metadata`.

**Interpretation:** The existing chunk artifact is accessible and successfully parsed. All 228 records are available in memory through `raw_chunk_records`.

This step confirms successful loading and basic JSON parsing only. Full schema and data validation will be performed in the following steps.


## 1.2 Validate ChunkRecord Objects

The existing `ChunkRecord` Pydantic model is defined in `app/chunking/models.py`.

Instead of introducing a new schema, the original project model will be used to validate every loaded record.

The validation covers:

- Required field types and constraints.
- Non-empty identifiers and strategy information.
- Valid page numbers, chunk indexes, and character offsets.
- SHA-256 content hash format.
- Section path and metadata structures.
- Additional field rejection.
- Custom character-offset consistency validation.

Each raw JSON record will be passed to `ChunkRecord.model_validate()`.

Successfully validated objects will be stored in `validated_chunk_records`, while validation failures will be collected separately for inspection.

The original `raw_chunk_records` will remain unchanged.

No chunk rebuilding, correction, or modification will be performed.

````python
import sys
from pydantic import ValidationError

# --------------------------------------------------
# 1. Import the Original Project Model
# --------------------------------------------------

project_root_str = str(PROJECT_ROOT)

if project_root_str not in sys.path:
    sys.path.insert(0, project_root_str)

from app.chunking.models import ChunkRecord


# --------------------------------------------------
# 2. Validate Loaded Chunk Records
# --------------------------------------------------

validated_chunk_records = []
chunk_validation_errors = []

for record_index, record in enumerate(raw_chunk_records):
    try:
        validated_record = ChunkRecord.model_validate(record)

        validated_chunk_records.append(validated_record)

    except ValidationError as exc:
        chunk_validation_errors.append(
            {
                "record_index": record_index,
                "chunk_id": (
                    record.get("chunk_id") if isinstance(record, dict) else None
                ),
                "errors": exc.errors(),
            }
        )


# --------------------------------------------------
# 3. Validation Summary
# --------------------------------------------------

total_records = len(raw_chunk_records)
valid_records = len(validated_chunk_records)
invalid_records = len(chunk_validation_errors)

validation_rate = (valid_records / total_records) * 100 if total_records else 0.0

print("=" * 60)
print("CHUNK RECORD VALIDATION SUMMARY")
print("=" * 60)

print("Total records:", total_records)
print("Valid records:", valid_records)
print("Invalid records:", invalid_records)
print(f"Validation rate: {validation_rate:.2f}%")

print(
    "Validated object type:",
    type(validated_chunk_records[0]).__name__
    if validated_chunk_records
    else "No valid records",
)

print("=" * 60)


# --------------------------------------------------
# 4. Inspect Validation Errors
# --------------------------------------------------

if chunk_validation_errors:
    print("\nVALIDATION ERRORS")

    for item in chunk_validation_errors[:5]:
        print("\nRecord index:", item["record_index"])
        print("Chunk ID:", item["chunk_id"])

        for error in item["errors"]:
            print(
                "Field:",
                error["loc"],
                "| Type:",
                error["type"],
                "| Message:",
                error["msg"],
            )

    if len(chunk_validation_errors) > 5:
        print(f"\nAdditional errors not displayed: {len(chunk_validation_errors) - 5}")

else:
    print("\nNo validation errors found.")


# --------------------------------------------------
# 5. Inspect First Validated Record
# --------------------------------------------------

if validated_chunk_records:
    first_chunk = validated_chunk_records[0]

    print("\nFIRST VALIDATED CHUNK")
    print("Chunk ID:", first_chunk.chunk_id)
    print("Document ID:", first_chunk.document_id)
    print("Strategy:", first_chunk.strategy)
    print("Content length:", len(first_chunk.content))
    print("Character offsets:", first_chunk.char_start, first_chunk.char_end)

print("=" * 60)
````

### Result Interpretation

All 228 loaded records passed validation using the original Pydantic `ChunkRecord` model.

Validation results:

- Total records: 228
- Valid records: 228
- Invalid records: 0
- Validation rate: 100%
- Validation errors: 0

All records satisfied the model's field constraints, including the custom character-offset consistency validator.

The validated records are available as `ChunkRecord` instances in `validated_chunk_records`.

**Interpretation:** The existing chunk records are fully compatible with the original project data model. No schema violations or character-offset inconsistencies were detected.

## 1.4 Build Document Metadata Lookup

A document metadata lookup will be constructed using `document_id` as the dictionary key.

The lookup will provide direct access to the original document records when resolving chunk provenance and preparing document-level metadata for the embedding pipeline.

Before constructing the lookup, the document identifiers will be inspected for:

- Missing `document_id` fields.
- Invalid or empty identifiers.
- Duplicate document identifiers.

The resulting dictionary will be stored in `document_metadata_lookup`.

Each value will retain the original document record without modifying its contents.

No chunk-to-document relationship validation is performed in this step.

````python
# --------------------------------------------------
# 1. Inspect Document Identifiers
# --------------------------------------------------

document_id_issues = []
document_ids = []

for record_index, record in enumerate(raw_document_records):
    if not isinstance(record, dict):
        document_id_issues.append(
            {"record_index": record_index, "issue": "Record is not a dictionary"}
        )
        continue

    document_id = record.get("document_id")

    if not isinstance(document_id, str) or not document_id.strip():
        document_id_issues.append(
            {"record_index": record_index, "issue": "Missing or invalid document_id"}
        )
        continue

    document_ids.append(document_id)


# --------------------------------------------------
# 2. Detect Duplicate Document IDs
# --------------------------------------------------

seen_document_ids = set()
duplicate_document_ids = []

for document_id in document_ids:
    if document_id in seen_document_ids:
        duplicate_document_ids.append(document_id)
    else:
        seen_document_ids.add(document_id)


# --------------------------------------------------
# 3. Build Lookup
# --------------------------------------------------

if document_id_issues or duplicate_document_ids:
    raise ValueError(
        "Document metadata lookup cannot be built safely.\n"
        f"Identifier issues: {document_id_issues}\n"
        f"Duplicate IDs: {duplicate_document_ids}"
    )

document_metadata_lookup = {
    record["document_id"]: record for record in raw_document_records
}


# --------------------------------------------------
# 4. Lookup Summary
# --------------------------------------------------

print("=" * 55)
print("DOCUMENT METADATA LOOKUP SUMMARY")
print("=" * 55)

print("Total document records:", len(raw_document_records))
print("Valid document IDs:", len(document_ids))
print("Unique document IDs:", len(seen_document_ids))
print("Duplicate document IDs:", len(duplicate_document_ids))
print("Identifier issues:", len(document_id_issues))
print("Lookup entries:", len(document_metadata_lookup))

if document_metadata_lookup:
    sample_id = next(iter(document_metadata_lookup))
    sample_record = document_metadata_lookup[sample_id]

    print("\nSample lookup:")
    print("Document ID:", sample_id)
    print("File name:", sample_record.get("file_name"))
    print("File type:", sample_record.get("file_type"))

print("=" * 55)
````

### Result Interpretation

The document metadata lookup was constructed successfully using `document_id` as the dictionary key.

Validation results:

- Total document records: 6
- Valid document IDs: 6
- Unique document IDs: 6
- Duplicate document IDs: 0
- Identifier issues: 0
- Lookup entries: 6

A sample lookup successfully resolved the document ID `5eaa2f3a4137bbd8` to `README_03_Local_Runtime_Engineering.md`.

**Interpretation:** All six document records have valid and unique identifiers. The `document_metadata_lookup` dictionary is ready to provide direct access to the original document metadata during chunk provenance validation and subsequent embedding preparation.

The original document records remain unchanged.

## 1.5 Verify Document-to-Chunk Relationships

## This step verifies the referential integrity between the existing chunk records and their source documents.

Each validated chunk contains a `document_id` that must resolve to an existing entry in `document_metadata_lookup`.

The verification covers:

1. Counting chunks with valid document references.
2. Identifying chunks referencing missing document IDs.
3. Counting documents associated with at least one chunk.
4. Identifying documents without associated chunks.
5. Reporting the number of chunks belonging to each document.

A chunk-to-document relationship is considered valid when its `document_id` exists in the document metadata lookup.

No chunk or document records will be modified.


````python
# Verify that every chunk references an existing source document.

valid_chunk_references = []
missing_document_references = []

# Initialize a chunk counter for every known document.
chunks_per_document = {document_id: 0 for document_id in document_metadata_lookup}

# Check each chunk's document_id against the metadata lookup.
for chunk in validated_chunk_records:
    document_id = chunk.document_id

    if document_id in document_metadata_lookup:
        valid_chunk_references.append(chunk.chunk_id)
        chunks_per_document[document_id] += 1

    else:
        missing_document_references.append(
            {"chunk_id": chunk.chunk_id, "document_id": document_id}
        )

# Separate documents with chunks from documents without chunks.
documents_with_chunks = [
    document_id for document_id, count in chunks_per_document.items() if count > 0
]

documents_without_chunks = [
    document_id for document_id, count in chunks_per_document.items() if count == 0
]

# Print relationship validation results.
print("=" * 60)
print("DOCUMENT-TO-CHUNK RELATIONSHIP SUMMARY")
print("=" * 60)

print("Total chunks:", len(validated_chunk_records))
print("Total documents:", len(document_metadata_lookup))
print("Chunks with valid references:", len(valid_chunk_references))
print("Chunks with missing references:", len(missing_document_references))
print("Documents with chunks:", len(documents_with_chunks))
print("Documents without chunks:", len(documents_without_chunks))

# Display the chunk distribution across source documents.
print("\nChunks per document:")

for document_id, chunk_count in chunks_per_document.items():
    file_name = document_metadata_lookup[document_id].get("file_name", "Unknown")

    print(f"- {file_name}: {chunk_count} chunks")

# Display missing references, if any.
if missing_document_references:
    print("\nMISSING DOCUMENT REFERENCES")

    for item in missing_document_references[:10]:
        print(item)

else:
    print("\nNo missing document references found.")

# Evaluate referential integrity without treating empty documents as errors.
relationships_valid = (
    len(valid_chunk_references) == len(validated_chunk_records)
    and len(missing_document_references) == 0
)

print("\nRelationship validation:", "PASS" if relationships_valid else "CHECK REQUIRED")

print("=" * 60)
````

### Result Interpretation

The document-to-chunk relationship verification completed successfully.

Validation results:

- Total chunks: 228
- Total documents: 6
- Chunks with valid document references: 228
- Chunks with missing document references: 0

## 2.1 Embedding Models vs Language Models

### Language Models (LLMs)

Language Models are primarily designed to understand and generate natural language.

Given a text prompt, an LLM processes the input and generates a sequence of output tokens.

Examples include GPT, Qwen, Llama, and Gemma.

### Embedding Models

Embedding Models convert input text into fixed-dimensional numerical vectors called embeddings.

These vectors represent semantic information and can be compared using similarity metrics such as cosine similarity.

Embedding Models are commonly used for:

- Semantic Search
- Dense Retrieval
- Document Retrieval
- Semantic Similarity
- Clustering

### Key Differences

| Comparison | Language Model | Embedding Model |
|---|---|---|
| Primary Purpose | Text generation | Vector representation |
| Main Output | Generated tokens | Numerical vectors |
| Semantic Search | Not its primary role | Core use case |
| Text Generation | Yes | No, not its primary function |
| Typical RAG Role | Answer generation | Document retrieval |

### Their Roles in a RAG System

A typical RAG pipeline uses both model types:

1. The Embedding Model converts the user's query into a numerical vector.
2. The Vector Database retrieves semantically similar document chunks.
3. The retrieved chunks are provided as context to the Language Model.
4. The Language Model generates the final response using the retrieved information.

**Key Takeaway:** Embedding Models transform text into numerical representations for retrieval, while Language Models generate natural language responses.

Both components serve different but complementary roles in a RAG architecture.

## 2.2 Dense Embeddings Concept

### What Are Dense Embeddings?

A Dense Embedding is a fixed-dimensional numerical representation of an input text.

It consists of a vector containing floating-point values that collectively encode semantic information learned by the embedding model.

For example:

```python
[0.12, -0.34, 0.56, 0.08, -0.21, ...]
```

Each dimension contributes to the overall representation. Individual dimensions do not necessarily correspond to directly interpretable concepts.

### Why Are They Called Dense?

Dense vectors contain values across most of their dimensions, unlike sparse representations, which contain mostly zero values.

| Property | Dense Embeddings | Sparse Representations |
|---|---|---|
| Representation | Values across most dimensions | Mostly zero values |
| Typical Dimensions | Fixed-dimensional | Often high-dimensional |
| Information Distribution | Distributed across dimensions | Concentrated in selected dimensions |
| Common Applications | Semantic Search | Lexical and Keyword Search |

### Semantic Representation

Dense Embedding models are trained to produce numerical representations that support semantic comparison.

Texts with related meanings can have more similar representations, even when they use different words.

For example:

- "A doctor treats patients in a hospital."
- "A physician provides medical care to people."

These sentences express related meanings despite their different wording.

### Embedding Space

An embedding space is a mathematical space in which each text is represented as a vector.

For `multilingual-e5-base`, each embedding is expected to contain 768 dimensions.

The model's representations can be compared using similarity metrics to identify semantically related texts.

### Cosine Similarity

Cosine similarity measures the angular similarity between two vectors.

Its value ranges from -1 to 1:

- 1: Same direction.
- 0: Orthogonal vectors.
- -1: Opposite directions.

The similarity score is used to compare query embeddings with document embeddings during semantic retrieval.

### Role in the RAG Pipeline

In this project:

1. Each of the 228 document chunks will be converted into a dense embedding.
2. The resulting vectors will be stored in ChromaDB.
3. A user query will be converted into an embedding using the same model.
4. ChromaDB will retrieve semantically similar document chunks.
5. The retrieved content will be passed to the Language Model as context.

**Key Takeaway:** Dense Embeddings provide numerical representations that allow a RAG system to retrieve semantically related information rather than relying exclusively on exact keyword matching.

The Transformer produces a contextual representation for each input token.

## 2.6 Maximum Sequence Length

### Definition

Maximum Sequence Length is the maximum number of tokens that an embedding model can process in a single input sequence.

For `intfloat/multilingual-e5-base`, the maximum sequence length is 512 tokens.

### Tokens vs. Words vs. Characters

These concepts represent different units of text:

| Concept | Meaning |
|---|---|
| Characters | Individual letters, symbols, and spaces |
| Words | Linguistic words |
| Tokens | Units generated by the tokenizer |

A single word may produce one or multiple tokens depending on the tokenizer and language.

Therefore, token counts cannot be reliably inferred from character or word counts.

### Input Truncation

When an input exceeds the model's maximum sequence length, the tokenizer can truncate the sequence to fit the configured limit.

The official model example uses:

```python
tokenizer(
    input_texts,
    max_length=512,
    padding=True,
    truncation=True,
    return_tensors="pt"
)
```

With truncation enabled, tokens exceeding the configured maximum length are discarded.

### Impact on RAG Retrieval

Truncation may remove relevant information from the end of a document chunk.

This can produce an embedding that does not represent the complete original chunk.

Potential consequences include:

- Information loss.
- Incomplete semantic representations.
- Reduced retrieval quality.
- Missing relevant context during retrieval.

### Chunk Size vs. Maximum Sequence Length

| Parameter | Meaning |
|---|---|
| Chunk Size | Amount of text produced by the document chunking process |
| Maximum Sequence Length | Maximum number of tokens processed by the embedding model |

These parameters are related but are not interchangeable.

Each document chunk must be evaluated using the embedding model's tokenizer.

### Special Tokens and Prefixes

The tokenizer may automatically add special tokens such as:

- `<s>`: Beginning of sequence.
- `</s>`: End of sequence.

E5 retrieval inputs also use task-specific prefixes:

- `passage:` for document passages.

## 2.7 Query and Passage Prefixes

### What Are Query and Passage Prefixes?

Prefixes are short text identifiers added to the beginning of input text before it is passed to the embedding model.

The multilingual E5 model uses task-specific prefixes during training to distinguish between queries and document passages.

### Supported Prefixes

| Prefix | Purpose | Usage |
|---|---|---|
| `query:` | Identifies a search query | User questions |
| `passage:` | Identifies a document passage | Document chunks |

### Asymmetric Retrieval

For document retrieval tasks, the model uses different prefixes for queries and document passages.

**Query Example:**

```text
query: How does the RAG pipeline retrieve documents?
```

**Passage Example:**

```text
passage: The retrieval component searches the vector database for semantically relevant document chunks.
```

The model generates embeddings for both inputs, which can then be compared using a similarity metric.

### Why Prefixes Matter

The E5 model was trained using these prefixes.

Omitting the appropriate prefixes can reduce retrieval performance.

Prefixes are part of the input text and provide task-related information to the model.

### Prefix Selection by Task

| Task | Prefix |
|---|---|
| Asymmetric document retrieval | `query:` for queries and `passage:` for documents |
| Semantic similarity | `query:` for both texts |
| Bitext mining | `query:` for both texts |
| Paraphrase retrieval | `query:` for both texts |
| Clustering | `query:` |

## 2.8 Model Revision and Reproducibility

### What Is Model Revision?

A model revision identifies a specific version of a model repository.

A Hugging Face model repository can contain multiple revisions as files and configurations evolve over time.

Using only a model identifier does not explicitly pin the model to a particular repository snapshot.

### The Main Branch

When a model is loaded without specifying a revision, the default revision is generally `main`.

The `main` branch points to the repository's current default revision and may change over time.

Consequently, loading the same model identifier at different dates does not necessarily guarantee identical model files.

### Commit Hash

A Git commit hash uniquely identifies a specific repository revision.

Instead of relying on the moving `main` branch, a fixed commit hash can be provided during model loading.

Example:

```python
model = SentenceTransformer(
    "intfloat/multilingual-e5-base",
    revision="COMMIT_HASH"
)
```

This allows the project to retrieve a specific model snapshot.

### Why Reproducibility Matters in RAG

Document and query embeddings must be generated using compatible model configurations.

Using different model revisions for document and query embeddings may produce incompatible vector representations, even when both revisions have the same model identifier and embedding dimension.

If the embedding model revision changes, document embeddings should be regenerated using the new revision.

````python
import hashlib
from pathlib import Path

# Define the required local model artifacts.
required_artifacts = [
    "sentencepiece.bpe.model",
    "tokenizer.json",
    "tokenizer_config.json",
    "special_tokens_map.json",
    "config.json",
    "modules.json",
    "1_Pooling/config.json",
    "model.safetensors",
]


# Define a function to calculate SHA-256 incrementally.
def calculate_sha256(file_path, chunk_size=1024 * 1024):
    sha256 = hashlib.sha256()

    with file_path.open("rb") as file:
        while True:
            chunk = file.read(chunk_size)

            if not chunk:
                break

            sha256.update(chunk)

    return sha256.hexdigest()


# Inspect each required artifact.
artifact_results = []

print("Tokenizer and Runtime Files Validation")
print("=" * 65)

for relative_path in required_artifacts:
    file_path = local_snapshot_dir / relative_path

    exists = file_path.is_file()
    size_bytes = file_path.stat().st_size if exists else 0
    non_empty = exists and size_bytes > 0

    readable = False
    sha256_hash = None
    error_message = None

    if non_empty:
        try:
            sha256_hash = calculate_sha256(file_path)
            readable = True
        except OSError as error:
            error_message = str(error)

    result = {
        "file": relative_path,
        "exists": exists,
        "size_bytes": size_bytes,
        "non_empty": non_empty,
        "readable": readable,
        "sha256": sha256_hash,
        "error": error_message,
    }

    artifact_results.append(result)

    print(f"\nFile: {relative_path}")
    print("Exists:", exists)
    print("Size (bytes):", size_bytes)
    print("Non-empty:", non_empty)
    print("Readable:", readable)
    print("SHA-256:", sha256_hash if sha256_hash else "Unavailable")

    if error_message:
        print("Error:", error_message)

# Evaluate the complete artifact inventory.
all_exist = all(item["exists"] for item in artifact_results)
all_non_empty = all(item["non_empty"] for item in artifact_results)
all_readable = all(item["readable"] for item in artifact_results)
all_hashed = all(item["sha256"] is not None for item in artifact_results)

# Display validation summary.
print("\nValidation Summary")
print("=" * 65)

print(f"Total Required Files: {len(required_artifacts)}")
print(f"Files Found: {sum(item['exists'] for item in artifact_results)}")
print(f"Non-empty Files: {sum(item['non_empty'] for item in artifact_results)}")
print(f"Readable Files: {sum(item['readable'] for item in artifact_results)}")
print(
    f"SHA-256 Generated: {sum(item['sha256'] is not None for item in artifact_results)}"
)

print("\nValidation Results")
print("File Existence:", "PASS" if all_exist else "FAIL")
print("Non-empty Files:", "PASS" if all_non_empty else "FAIL")
print("File Readability:", "PASS" if all_readable else "FAIL")
print("SHA-256 Generation:", "PASS" if all_hashed else "FAIL")

# Enforce all validation requirements.
assert all_exist, "One or more required artifacts are missing."
assert all_non_empty, "One or more required artifacts are empty."
assert all_readable, "One or more required artifacts cannot be read."
assert all_hashed, "SHA-256 generation failed for one or more artifacts."

print("\nOverall Status: PASS")
````

### 2.10.12 Query and Passage Embedding Compatibility

**Objective**

Validate numerical compatibility between query and passage embeddings and verify that cosine similarity can be calculated correctly between the two embedding groups.

**Similarity Metric**

Cosine similarity measures the angular similarity between two vectors:
$$
\frac{q \cdot p}{|q|\_2 |p|\_2}
$$

Since both embedding groups are L2-normalized, cosine similarity is equivalent to their dot product, subject to floating-point precision.

**Validation Scope**

1. Verify that query and passage embeddings have compatible dimensions.
2. Verify that both groups have matching data types and devices.
3. Calculate the pairwise cosine similarity matrix.
4. Validate the similarity matrix shape and numerical finiteness.
5. Verify that all similarity scores fall within the valid cosine similarity range, allowing a small numerical tolerance.
6. Inspect corresponding query-passage similarity scores.

**Experimental Constraints**

- Reuse the query and passage embeddings generated in Section 2.10.11.
- Do not regenerate or modify the embeddings.
- Do not assume that corresponding query-passage pairs must exceed a fixed similarity threshold.
- Do not interpret numerical compatibility as proof of retrieval accuracy.

**Expected Outcome**

Confirm that query and passage embeddings are mathematically compatible and that a valid pairwise cosine similarity matrix can be calculated.

````python
import torch

# Numerical tolerance
similarity_tolerance = 1e-5

# 1. Validate embedding compatibility
query_dimension = query_embeddings.shape[-1]
passage_dimension = passage_embeddings.shape[-1]

dimension_status = query_dimension == passage_dimension
dtype_status = query_embeddings.dtype == passage_embeddings.dtype
device_status = query_embeddings.device == passage_embeddings.device

print("Embedding Compatibility")
print(f"Query Shape: {tuple(query_embeddings.shape)}")
print(f"Passage Shape: {tuple(passage_embeddings.shape)}")
print(f"Query Dimension: {query_dimension}")
print(f"Passage Dimension: {passage_dimension}")
print(f"Dimension Compatibility: {'PASS' if dimension_status else 'FAIL'}")
print(f"Data Type Compatibility: {'PASS' if dtype_status else 'FAIL'}")
print(f"Device Compatibility: {'PASS' if device_status else 'FAIL'}")

# 2. Calculate pairwise cosine similarity
similarity_matrix = torch.matmul(query_embeddings, passage_embeddings.T)

# 3. Validate similarity matrix
expected_similarity_shape = (query_embeddings.shape[0], passage_embeddings.shape[0])

shape_status = tuple(similarity_matrix.shape) == expected_similarity_shape

finite_status = torch.isfinite(similarity_matrix).all().item()

range_status = (similarity_matrix >= -1.0 - similarity_tolerance).all().item() and (
    similarity_matrix <= 1.0 + similarity_tolerance
).all().item()

print("\nCosine Similarity Matrix")
print(f"Expected Shape: {expected_similarity_shape}")
print(f"Actual Shape: {tuple(similarity_matrix.shape)}")
print(f"Similarity Matrix:\n{similarity_matrix.cpu()}")
print(f"Shape Check: {'PASS' if shape_status else 'FAIL'}")
print(f"Finite Values: {'PASS' if finite_status else 'FAIL'}")
print(f"Valid Similarity Range: {'PASS' if range_status else 'FAIL'}")

# 4. Inspect corresponding query-passage pairs
diagonal_scores = torch.diagonal(similarity_matrix)

print("\nCorresponding Query-Passage Similarity")
for index, score in enumerate(diagonal_scores.cpu().tolist()):
    print(f"Pair {index + 1}: {score:.6f}")

# 5. Overall validation
checks = {
    "Dimension Compatibility": dimension_status,
    "Data Type Compatibility": dtype_status,
    "Device Compatibility": device_status,
    "Similarity Matrix Shape": shape_status,
    "Finite Similarity Values": finite_status,
    "Valid Similarity Range": range_status,
}

print("\nCompatibility Validation Summary")

for check_name, passed in checks.items():
    print(f"{check_name}: {'PASS' if passed else 'FAIL'}")

print(f"\nPassed Checks: {sum(checks.values())}/{len(checks)}")
print("Overall Status:", "PASS" if all(checks.values()) else "FAIL")
````
